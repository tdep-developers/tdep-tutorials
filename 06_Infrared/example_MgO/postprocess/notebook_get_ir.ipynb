```python
import numpy as np
import pandas as pd
import xarray as xr
from ase.io import read
from matplotlib import pyplot as plt
from tdeptools.kramers_kronig import get_kk_real_from_imag_uneven as get_kk_real

to_invcm = 33.356
```

```python
def read_dataset(file: str, group="anharmonic") -> xr.Dataset:
    nq = int(xr.load_dataarray(file, group="qmesh")[0])
    ds = xr.load_dataset(file, group=group)
    ds.attrs.update({"nq": nq})
    return ds
```

```python
# Read the structure
atoms = read("../infile.ucposcar", format="vasp")

# Read the spectral function
ds = read_dataset("../lineshape/outfile.phonon_self_energy.hdf5")

# Read the mode oscillator strengths
df_intensity = pd.read_csv("../lineshape/outfile.ir_intensity.csv", index_col="mode")

# Read the electronic contribution to the dielectric tensor, pick xx component (why?)
data_loto = np.loadtxt("../infile.lotosplitting").reshape(-1, 3, 3)
eps_inf = data_loto[0][0, 0]

df_intensity
```

```python
# spectral function = imaginary part of G:

# Which mode are we interested in?
s_mode = 4

# angular frequency:
ws = df_intensity.frequency.iloc[s_mode]

Z = df_intensity.Z_x.iloc[s_mode]

S = Z ** 2 * df_intensity.prefactor.iloc[s_mode]

G_imag = np.pi * ds.spectralfunction_per_mode[s_mode]

# KK to get real
G_real = get_kk_real(xs=ds.frequency, ys=G_imag)

# the full G
G = G_real + 1.0j * G_imag


pref = ws / 2  # makes G(0) -> ~1

# full dielectric function
epsilon = eps_inf + S * pref * G


# Optical functions, compare Eq. (1) in Fugallo2018:

## refraction index
n = np.sqrt(0.5 * (abs(epsilon) + epsilon.real))

## reflectivity r
epsilon_sqrt = np.sqrt(epsilon)
r = abs((epsilon_sqrt - 1) / (epsilon_sqrt + 1)) ** 2

## Absorption
d = 1e-1  # sample thickness, figure out unit
A = epsilon.imag / n
a = A * d
t = np.exp(-a)

# transmission
T = t * (1 - r) ** 2 / (1 - r ** 2 * t ** 2)
R = r * (1 + t * T)
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(nrows=3, sharex=True)

# frequency axis
x = ds.frequency * to_invcm
plt.suptitle('Dielectric quantities')
ax1.plot(x, epsilon.imag)
ax1.plot(x, epsilon.real)
ax1.plot(x, (1 / epsilon).imag)
ax1.legend(["Im $\epsilon$", "Re $\epsilon$", "Im $1/\epsilon$"])
ax1.set_ylim(-10, 20)

ax2.plot(x, R)
ax2.plot(x, T)
ax2.plot(x, 1 - t)
ax2.legend(["R", "T", "A"])

ax3.plot(x, n)
ax3.plot(x, epsilon.imag / n)
ax3.set_ylim(0, 10)
ax3.legend(["n", "$\kappa$"])
ax3.set_xlim(0, 2000)

plt.xlabel('Wavenumber (cm$^{-1}$)')

for ax in (ax1, ax2, ax3):
    ax.axvline(ws * to_invcm, color="#313131", ls="--")
```

```python
# spectral function = imaginary part of G:

# Which mode are we interested in?

ws = []
Z = []
S = []
G = []
epsilon = []
n = []
r = []
d = []
T = []
R = []
t = []


n_modes = len(df_intensity.frequency.iloc[:])
for s_mode in range(n_modes):
    # angular frequency:
    ws.append(df_intensity.frequency.iloc[s_mode])
    
    Z.append(df_intensity.Z_x.iloc[s_mode])
    
    S.append(Z[s_mode] ** 2 * df_intensity.prefactor.iloc[s_mode])
    
    G_imag = np.pi * ds.spectralfunction_per_mode[s_mode]
    
    # KK to get real
    G_real = get_kk_real(xs=ds.frequency, ys=G_imag)
    
    # the full G
    G.append(G_real + 1.0j * G_imag)
    
    
    pref = ws[s_mode] / 2  # makes G(0) -> ~1
    
    # full dielectric function
    epsilon.append(eps_inf + S[s_mode] * pref * G[s_mode])
    
    
    # Optical functions, compare Eq. (1) in Fugallo2018:
    
    ## refraction index
    n.append(np.sqrt(0.5 * (abs(epsilon[s_mode]) + epsilon[s_mode].real)))
    
    ## reflectivity r
    epsilon_sqrt = np.sqrt(epsilon[s_mode])
    r = abs((epsilon_sqrt - 1) / (epsilon_sqrt + 1)) ** 2
    
    ## Absorption
    d = 1e-1  # sample thickness, figure out unit
    A = epsilon[s_mode].imag / n[s_mode]
    a = A * d
    t.append(np.exp(-a))
    
    # transmission
    T.append(t[s_mode] * (1 - r) ** 2 / (1 - r ** 2 * t[s_mode] ** 2))
    R.append(r * (1 + t[s_mode] * T[s_mode]))
```

```python
fig, (ax1, ax2, ax3, ax4) = plt.subplots(nrows=4, sharex=True)
# frequency axis
x = ds.frequency * to_invcm

s_mode = 4
ax1.plot(x, epsilon[s_mode].imag)
ax1.plot(x, epsilon[s_mode].real)
ax1.plot(x, (1 / epsilon[s_mode]).imag)
ax2.plot(x, R[s_mode])
ax2.plot(x, T[s_mode])
ax2.plot(x, 1 - t[s_mode])
ax3.plot(x, n[s_mode])
ax3.plot(x, epsilon[s_mode].imag / n[s_mode])

ax1.set_ylim(-10, 20)
ax3.set_ylim(0, 10)

ax1.legend(["Im $\epsilon$", "Re $\epsilon$", "Im $1/\epsilon$"])
ax2.legend(["R", "T", "A"])
ax3.legend(["n", "$\kappa$"])
ax3.set_xlim(0, 2000)

for ax in (ax1, ax2, ax3):
    ax.axvline(ws[s_mode] * to_invcm, color="#313131", ls="--")

# TOTAL ABSORPTION SPECTRUM
ax4.plot(x, (1-np.array(t)).sum(axis=0))
```
